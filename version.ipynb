```python
import subprocess
from pathlib import Path

def get_git_version(path: str | Path = ".") -> str | None:
    """
    Returns a git version string if available, otherwise None.

    Examples:
        v1.2.0
        v1.2.0-4-gabc1234
        abc1234
    """

    path = Path(path)

    try:
        result = subprocess.run(
            [
                "git",
                "describe",
                "--tags",
                "--always",
                "--dirty",
            ],
            cwd=path,
            capture_output=True,
            text=True,
            check=True,
        )

        version = result.stdout.strip()

        return version or None

    except (
        subprocess.CalledProcessError,
        FileNotFoundError,
        OSError,
    ):
        return None
    
get_git_version()
```
